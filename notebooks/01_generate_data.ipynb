## 🎯 What This Notebook Does
1. Uses the `Faker` library to generate realistic employee records
2. Generates datasets of varying sizes: **1K, 10K, 100K, and 1 Million** records
3. Stores the final 1M dataset to `data/employees.csv`
4. Measures and reports generation time for each dataset size

## 📋 Employee Record Schema
| Field  | Type   | Description                  |
|--------|--------|------------------------------|
| name   | string | Full name (First + Last)     |
| email  | string | Unique email address         |
| city   | string | City of residence            |
| salary | float  | Annual salary (30K–150K USD) |

```python
from faker import Faker
import pandas as pd
import random

# Initialize Faker
fake = Faker()

# Number of records
TOTAL_RECORDS = 10000

# Store data
data = []

# Generate fake employee data
for _ in range(TOTAL_RECORDS):

    employee = {
        "name": fake.name(),
        "email": fake.email(),
        "city": fake.city(),
        "salary": random.randint(30000, 200000)
    }

    data.append(employee)

# Convert to DataFrame
df = pd.DataFrame(data)

# Save CSV
df.to_csv("employees.csv", index=False)

print("✅ employees.csv generated successfully")
print(f"📊 Total Records: {TOTAL_RECORDS}")
```
